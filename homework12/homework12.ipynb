## 2 Plotting
Make a scatter plot of **exoplanet Earth radius vs. equilibrium temperatures**. Include units, x/y axis labels, and a legend.

```python
import matplotlib.pyplot as plt

#creating figure and plotting data
plt.figure(figsize=(7, 5))
plt.scatter(planet_data["pl_eqt"], planet_data["pl_rade"], label="Exoplanets")
plt.legend()

#axes, titles, and units
plt.xlabel("Equilibrium Temperatures ($K$)")
plt.ylabel("Exoplanet Radius ($Earth Radii$)")
plt.title("Exoplanet Radius vs. Equilibrium Temperatures")
```

## 3 A "Cool" Planet
What planet has the **lowest** equilibrium temperature? Print the planet name and temperature with your `planet_data` variable.

```python
#calculate index of lowest equilibrium temperature
lowest_eqt_index = planet_data["pl_eqt"].idxmin()

#print out information about the planet with that index
print(planet_data.loc[lowest_eqt_index])
```

Make a new variable that only contains the information for that planet. \
\
*Hint: Use the `.iloc[[#]]` function. Replace # with the proper index.*

```python
lowest_eqt_planet = planet_data.iloc[[lowest_eqt_index]]
print(lowest_eqt_planet)
```

Print the number of columns in the row. 

```python
print("The number of columns is " + str(len(lowest_eqt_planet.columns)) + ".")
```

Make a new variable that contains only the columns WITHOUT `Nan` values. *Hint: You can use the function `.dropna(axis=1)`.*

```python
#drop the column with NaN values
no_nan = lowest_eqt_planet.dropna(axis=1)
print(no_nan)
```
